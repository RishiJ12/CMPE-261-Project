## 2. Panel cropping helpers
Storyboard pages often hold several panels. Use whichever approach fits your files. Crop first, resize later.

```python
def load_rgb(path):
    return Image.open(path).convert('RGB')   # drops alpha channels that break things downstream

def split_grid(img, rows, cols, margin=0):
    """Evenly split a page into rows x cols panels (reading order)."""
    w, h = img.size
    pw, ph = w // cols, h // rows
    panels = []
    for r in range(rows):
        for c in range(cols):
            box = (c*pw + margin, r*ph + margin, (c+1)*pw - margin, (r+1)*ph - margin)
            panels.append(img.crop(box))
    return panels

def detect_panels(img, min_w=150, min_h=100, thresh_val=200, row_tol=50):
    """Rough auto-detection via contours. Works only if panels have visible borders.
    Always check the result visually."""
    gray = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2GRAY)
    _, thresh = cv2.threshold(gray, thresh_val, 255, cv2.THRESH_BINARY_INV)
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    boxes = []
    for c in contours:
        x, y, w, h = cv2.boundingRect(c)
        if w > min_w and h > min_h:
            boxes.append((x, y, x+w, y+h))
    boxes.sort(key=lambda b: (b[1] // row_tol, b[0]))
    return [img.crop(b) for b in boxes], boxes

def show(images, titles=None, cols=4, size=4):
    rows = (len(images) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols*size, rows*size))
    axes = np.array(axes).reshape(-1)
    for ax in axes: ax.axis('off')
    for i, im in enumerate(images):
        axes[i].imshow(im)
        if titles: axes[i].set_title(titles[i], fontsize=9)
    plt.tight_layout(); plt.show()

def save_sequence(panel_images, seq_name):
    paths = []
    for i, p in enumerate(panel_images, 1):
        path = os.path.join(PANEL_DIR, f'{seq_name}_panel{i:03d}.png')
        p.save(path); paths.append(path)
    return paths
```

## 3. Conditioning map extraction
ControlNet's canny model expects **white lines on black**. Storyboards are usually dark lines on white, so try both modes and compare.

```python
def make_condition(img, mode='canny', size=512, low=100, high=200):
    img = img.resize((size, size))   # same resize for every panel and every method
    gray = np.array(img.convert('L'))
    if mode == 'canny':
        cond = cv2.Canny(gray, low, high)
    elif mode == 'invert':
        cond = 255 - gray            # dark lines on white -> white lines on black
    else:
        raise ValueError(mode)
    return Image.fromarray(cond).convert('RGB')

# Quick comparison on one panel (run after you've saved some panels)
panel_paths = sorted(glob.glob(os.path.join(PANEL_DIR, '*.png')))
if panel_paths:
    test = load_rgb(panel_paths[0])
    show([test.resize((512, 512)),
          make_condition(test, 'canny'),
          make_condition(test, 'invert')],
         titles=['panel', 'canny', 'invert'], cols=3)
```

## 4. Load ControlNet

```python
from diffusers import StableDiffusionControlNetPipeline, ControlNetModel, UniPCMultistepScheduler

controlnet = ControlNetModel.from_pretrained(CONFIG['controlnet'], torch_dtype=torch.float16)
pipe = StableDiffusionControlNetPipeline.from_pretrained(
    CONFIG['base_model'], controlnet=controlnet, torch_dtype=torch.float16
).to('cuda')
pipe.scheduler = UniPCMultistepScheduler.from_config(pipe.scheduler.config)
pipe.set_progress_bar_config(disable=True)
```

## 5. Sequence generation loop
Each sequence is a list of panel files plus one text prompt per panel. The seed is reset for every panel so differences between methods come from the method, not the noise. Outputs and the exact config are saved together.

```python
def run_sequence(seq_name, panel_files, prompts, method='controlnet_baseline', cfg=CONFIG):
    assert len(panel_files) == len(prompts), 'need one prompt per panel'
    out_dir = os.path.join(OUTPUT_DIR, method, seq_name)
    os.makedirs(out_dir, exist_ok=True)

    results = []
    for i, (pf, prompt) in enumerate(zip(panel_files, prompts), 1):
        cond = make_condition(load_rgb(pf), cfg['cond_mode'], cfg['resolution'],
                              cfg['canny_low'], cfg['canny_high'])
        gen = torch.Generator('cuda').manual_seed(cfg['seed'])
        img = pipe(
            prompt=f"{prompt}, {cfg['prompt_suffix']}",
            negative_prompt=cfg['negative_prompt'],
            image=cond,
            num_inference_steps=cfg['steps'],
            guidance_scale=cfg['guidance_scale'],
            controlnet_conditioning_scale=cfg['controlnet_conditioning_scale'],
            generator=gen,
        ).images[0]
        img.save(os.path.join(out_dir, f'panel{i:02d}.png'))
        cond.save(os.path.join(out_dir, f'panel{i:02d}_cond.png'))
        results.append(img)

    record = {"method": method, "sequence": seq_name, "panel_files": panel_files,
              "prompts": prompts, "config": cfg, "timestamp": time.strftime('%Y-%m-%d %H:%M:%S'),
              "torch": torch.__version__}
    with open(os.path.join(out_dir, 'run_config.json'), 'w') as f:
        json.dump(record, f, indent=2)
    return results

# --- Define a sequence: same character described the same way in every prompt ---
seq_name = 'seq01'
panel_files = sorted(glob.glob(os.path.join(PANEL_DIR, f'{seq_name}_panel*.png')))
CHAR = "a vampire with dark skin, white eye rings"
SETTING = "white background"
#6 panels
actions = [
    "jelly vaporizes as camera pans up",
    "vampire hands slowly come into screen, close-up",
    "hands rapidly push vaporized mist upwards, close-up",
    "camera pans upto vampire's face, close-up",
    "vampire inhales red jelly mist from right hand, close-up",
    "vampire inhales red jelly mist from left hand, close-up",
]
prompts = [f"{CHAR}, {a}, {SETTING}" for a in actions][:len(panel_files)]

if panel_files:
    frames = run_sequence(seq_name, panel_files, prompts)
```

## 6b. Consistent Self-Attention (adapted from StoryDiffusion)

How it works: all panels of a sequence are generated as one batch. In each self-attention layer, every panel's keys/values are extended with tokens randomly sampled from the *other* panels in the batch, so panels can "see" how the others render the character.

Notes:
- With classifier-free guidance the UNet batch is `[uncond panels; cond panels]`, so tokens are shared only within each half (`groups = 2`). The code sets this automatically from `guidance_scale`.
- `rate` is the fraction of the other panels' tokens that is sampled. I have not verified the official default, so treat it as a hyperparameter and sweep it (e.g. 0.25 / 0.5 / 0.75).
- If you run out of GPU memory, lower `max_tokens` (e.g. 1024) so sharing only happens in the lower-resolution layers, or generate fewer panels per batch. Use the same setting for every reported run.
- Turning the module off gives plain batched ControlNet generation, which should look nearly identical to your per-panel baseline. Check this first; it validates the batching path.

```python
from diffusers.models.attention_processor import AttnProcessor2_0

# Shared switchboard read by every processor
STATE = {"enabled": False, "groups": 2, "rate": 0.5, "max_tokens": 4096}

class ConsistentSelfAttnProcessor:
    def __init__(self):
        self.default = AttnProcessor2_0()

    def __call__(self, attn, hidden_states, encoder_hidden_states=None,
                 attention_mask=None, temb=None, *args, **kwargs):
        B, N, C = hidden_states.shape
        G = STATE["groups"]
        P = B // G                                    # panels per group
        # fall back to normal attention when sharing doesn't apply
        if (not STATE["enabled"]) or encoder_hidden_states is not None \
           or P < 2 or N > STATE["max_tokens"]:
            return self.default(attn, hidden_states, encoder_hidden_states,
                                attention_mask, temb, *args, **kwargs)
        x = hidden_states.reshape(G, P, N, C)
        ctx = []
        for p in range(P):
            others = torch.cat([x[:, q] for q in range(P) if q != p], dim=1)
            k = int(STATE["rate"] * others.shape[1])
            idx = torch.randperm(others.shape[1], device=x.device)[:k]
            ctx.append(torch.cat([x[:, p], others[:, idx]], dim=1))   # own + sampled tokens
        ctx = torch.stack(ctx, dim=1).reshape(B, N + k, C)
        return self.default(attn, hidden_states, ctx, attention_mask, temb, *args, **kwargs)

def install_csa(pipe):
    """Replace self-attention (attn1) processors; leave cross-attention (attn2) alone."""
    procs = {}
    for name in pipe.unet.attn_processors.keys():
        procs[name] = (ConsistentSelfAttnProcessor() if name.endswith("attn1.processor")
                       else AttnProcessor2_0())
    pipe.unet.set_attn_processor(procs)

install_csa(pipe)    # STATE['enabled'] is False by default, so earlier cells behave as before
```

```python
def warn_black(imgs, label):
    """The SD safety checker returns an all-black image when it flags a result."""
    for i, im in enumerate(imgs, 1):
        if np.array(im).max() < 5:
            print(f"WARNING [{label}] panel {i} is all black (safety checker flagged it). "
                  "Reword the prompt or change the setup consistently for ALL methods.")

def run_sequence_batch(seq_name, panel_files, prompts, method, use_csa,
                       cfg=CONFIG, rate=None, max_tokens=None):
    """Generate all panels of a sequence as ONE batch, with Consistent Self-Attention on/off."""
    assert len(panel_files) == len(prompts), 'need one prompt per panel'
    STATE["enabled"] = use_csa
    STATE["groups"] = 2 if cfg["guidance_scale"] > 1 else 1
    if rate is not None: STATE["rate"] = rate
    if max_tokens is not None: STATE["max_tokens"] = max_tokens

    out_dir = os.path.join(OUTPUT_DIR, method, seq_name)
    os.makedirs(out_dir, exist_ok=True)

    conds = [make_condition(load_rgb(pf), cfg['cond_mode'], cfg['resolution'],
                            cfg['canny_low'], cfg['canny_high']) for pf in panel_files]
    torch.manual_seed(cfg['seed'])        # makes the random token sampling reproducible
    gens = [torch.Generator('cuda').manual_seed(cfg['seed']) for _ in panel_files]  # same noise per panel, as in the baseline

    try:
        imgs = pipe(
            prompt=[f"{p}, {cfg['prompt_suffix']}" for p in prompts],
            negative_prompt=[cfg['negative_prompt']] * len(prompts),
            image=conds,
            num_inference_steps=cfg['steps'],
            guidance_scale=cfg['guidance_scale'],
            controlnet_conditioning_scale=cfg['controlnet_conditioning_scale'],
            generator=gens,
        ).images
    finally:
        record_state = dict(STATE)
        STATE["enabled"] = False          # always switch off so other cells aren't affected

    warn_black(imgs, method)
    for i, (im, cd) in enumerate(zip(imgs, conds), 1):
        im.save(os.path.join(out_dir, f'panel{i:02d}.png'))
        cd.save(os.path.join(out_dir, f'panel{i:02d}_cond.png'))
    record = {"method": method, "sequence": seq_name, "panel_files": panel_files,
              "prompts": prompts, "config": cfg, "csa_state": record_state,
              "timestamp": time.strftime('%Y-%m-%d %H:%M:%S'), "torch": torch.__version__}
    with open(os.path.join(out_dir, 'run_config.json'), 'w') as f:
        json.dump(record, f, indent=2)
    return imgs
```

```python
if panel_files:
    # 1) module OFF: plain batched ControlNet (should look ~like the per-panel baseline)
    frames_off = run_sequence_batch(seq_name, panel_files, prompts,
                                    'controlnet_batch_csa_off', use_csa=False)
    # 2) module ON: Consistent Self-Attention
    frames_on = run_sequence_batch(seq_name, panel_files, prompts,
                                   'storydiffusion_csa_on', use_csa=True)

    n = len(frames_on)
    show(frames_off + frames_on,
         titles=[f'CSA off {i+1}' for i in range(n)] + [f'CSA on {i+1}' for i in range(n)],
         cols=n)

# Optional: sweep the sampling rate
# for r in (0.25, 0.5, 0.75):
#     run_sequence_batch(seq_name, panel_files, prompts, f'storydiffusion_csa_rate{r}', True, rate=r)
```
